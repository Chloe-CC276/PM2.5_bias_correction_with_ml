Retrieval of AURN hourly PM2.5 observations

```python
import obsaq
```

```python


# UK geographic bounding box [lon_min, lon_max, lat_min, lat_max]
lon_min = -9
lon_max = 1.8
lat_min = 49
lat_max = 61
bounds=[lon_min, lon_max, lat_min, lat_max]

# Load AURN station metadata
meta = obsaq.meta()
site_table = meta.get_metadata('aurn')

# Select stations whose coordinates fall inside the study domain
final_sites = meta.get_site(bounds=bounds)
final_sites.head(5)

final_sites.to_csv("D:/Man/ERP/Data/AURN/aurn_metadata.csv")
```

```python
# Download hourly PM2.5 for the selected stations
merged_df = meta.download_sites(
    port="aurn",
    pollutant="PM2.5",
    start="2017-12-01",
    end="2018-12-02",
    output_dir="D:/Man/ERP/Data/AURN",
    download_mode="stream",     
    save_per_site=True,
    save_merged=True,
    add_site_id=True
)
```
